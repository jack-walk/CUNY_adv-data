```python
import pandas as pd
```

```python
#Manually set data types by column import
data_types = {
    "First Name":"str"
}

#Set aside which column imports will be dates
date_cols=["Fiscal Year", "Agency Start Date"]

#Load CSV and set preconditions
data = pd.read_csv(
    '../input/Citywide_Payroll_Data_(Fiscal_Year)_20261013.csv',
    low_memory=False,
    #Manually set data types
    dtype=data_types,
    #Parse date columns as dates
    parse_dates=date_cols
    )
len(data)
data.head()
```

```python
# logic operators return a boolean (true/false conditions)

# Return only the salaried employees by filtering for per Annum pay entries

data_reduced = data[data["Pay Basis"] == "per Annum"]

print(len(data_reduced))
print(len(data))

print(f"Rows in reduced dataset: {len(data_reduced)}. Rows in full dataset: {len(data)}.")
```

```python
# lambda lowkey means function -- initializes function
# one line of code to run through a whole dataset brrr

# Look at pay basis column values, lowercase it, 
# remove whitespace, and see if per annum is in that string. 
# Then put the true/false result in a new column, is_per_annum

# Value in Pay Basis column is being assigned to x

data["is_per_annum"] = data["Pay Basis"].apply(lambda x: "per annum" in str(x).lower())

data.head()
```

```python
#First need to convert to numeric bc my code is being weird

data["Base Salary"] = (
    data["Base Salary"]
    .astype(str)
    .str.replace(r"[$,]", "", regex=True)
    .str.strip()
)
data["Base Salary"] = pd.to_numeric(data["Base Salary"], errors="coerce")
```

```python
# Get median pay per job position by using groupby
# and applying same function to each job type

# Group by Title Description and identify its Base Salary column
# as the one we want to use for our calculations.
#Then take mean

data.groupby(
    ["Title Description"]
)["Base Salary"].median(

).reset_index(

).sort_values(
    by="Base Salary",
    ascending=False
)
```
