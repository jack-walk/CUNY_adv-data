```python
# Say you want to analyze the racial disparities in this dataset, broken down
# by race and by the larger category of non-white people.

# Write a line of code that takes a look at the universe of values in the race column.


# Get all values in column

race_values = data['race'].unique().tolist()
print(race_values)
```

```python
# Make the letters in the race column all lower-case

data["race_lowercase"] = data["race"].str.lower()
data.head()

race_lowercase_values = data['race_lowercase'].unique().tolist()
print(race_lowercase_values)
```

```python
# Create a new column called "poc" and use a lambda function to set 
# it equal to True or False based on the information in the race column.

# How many rows are POC and how many are White?

# Make a POC column and mark entries as "true" if they are POC
data["race_poc"] = data["race_lowercase"].apply(lambda x: str(x).lower() != "w")

# Count
data["race_poc"].value_counts(dropna=True)
```

Therefore, 5,771 entries are POC (or white and another value), and 4,659 are white.

```python
# I want to know which county had the most police killings but some counties in the U.S. 
# have the same name. 

# Create a column called country_state and populate it with the county and the state to 
# create unique pairs. 


# Now answer the question: which county had the most killings?

# Concatenate county and state data
data["county_state"] = data["county"] + ' ' + data["state"]

# Count entries per county_state value
data["county_state"].value_counts(dropna=True)
```

Therefore, Los Angeles County, CA had the most killings with 222.

```python
# Create a new column called "unarmed" and use a lambda function to set it to True/False based on the relevant info.

# Create a subset of the data for unarmed killings and store it in a new variable called unarmed_killings. 

# Which county had the most of these?


# Create new column
data["unarmed"] = data["armed_with"].apply(lambda x: str(x).lower() == "unarmed")


# Create new data subset
unarmed_killings = []

# Append w relevant data
for index, row in data.iterrows():
    if row["unarmed"] == True:
        unarmed_killings.append(row)

# Check work
unarmed_killings_df = pd.DataFrame(unarmed_killings)
unarmed_killings_df["armed_with"].value_counts(dropna=True)
```

```python
# We could have done that more smoothly using a groupby. So let's do that. 
# Group the original dataset by your new columns unarmed killings and county_state. 
# Use loc[] to isolate the rows where unarmed==True
# Sort from highest #killings to lowest

data.loc[data["unarmed"] == True].groupby(
    ["county_state"]
).size().reset_index(
    name="count"
).sort_values(
    by="count",
    ascending=False
)
```

```python
# Declare a list

women_only = []

# Append w relevant data
for index, row in data.iterrows():
    if row["gender"] == "female":
        women_only.append(row)

#Check work
women_only_count = pd.DataFrame(women_only)
women_only_count["gender"].value_counts(dropna=True)

# Return counts by race
women_only_count["race"].value_counts()
```

```python
# Based on returned values, store in variables
# (add +1 to Hispanic and Native bc one person is labeled N;H)
white_women_count = (women_only_count["race"].str.lower() == "w").sum()
black_women_count = (women_only_count["race"].str.lower() == "b").sum()
hisp_women_count = ((women_only_count["race"].str.lower() == "h").sum() + 1)
asian_women_count = (women_only_count["race"].str.lower() == "a").sum()
native_women_count = ((women_only_count["race"].str.lower() == "n").sum() + 1)
other_women_count = (women_only_count["race"].str.lower() == "o").sum()

print(white_women_count, black_women_count, hisp_women_count, asian_women_count, native_women_count, other_women_count)
```

```python
# Import U.S. women populations by race in 2024 (most recent year in dataset)
# Source: U.S. Census Bureau (https://www.census.gov/programs-surveys/popest.html)

white_women_pop = 128475000
black_women_pop = 24100000
hisp_women_pop = 31800000
asian_women_pop = 11900000
native_women_pop = 2400000

# Calculate number of killings per racial group

white_women_rate = white_women_count / white_women_pop
black_women_rate = black_women_count / black_women_pop
hisp_women_rate = hisp_women_count / hisp_women_pop
asian_women_rate = asian_women_count / asian_women_pop
native_women_rate = native_women_count / native_women_pop

# Store in a dataframe
dataWomenByRace = pd.DataFrame({
    "Race": ["White", "Black", "Hispanic", "Asian", "Native"],
    "Pop": [white_women_pop, black_women_pop, hisp_women_pop, asian_women_pop, native_women_pop],
    "Killings": [white_women_count, black_women_count, hisp_women_count, asian_women_count, native_women_count],
    "Rate": [white_women_rate, black_women_rate, hisp_women_rate, asian_women_rate, native_women_rate]
})

# Sort bty rate (commented out)
# dataWomenByRace.sort_values(by="Rate", ascending=False)

# Calculate how much more Black women are killed than white women
comparison = black_women_rate / white_women_rate

print(f"Between 2014 and 2025, Black women were killed by police at {comparison} times the rate of white women.")
```
