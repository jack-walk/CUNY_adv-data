### Let's scrape Project Gutenberg (!!!!)

https://www.gutenberg.org/ebooks/search/?sort_order=release_date

```python
import csv
from bs4 import BeautifulSoup
```

```python
books = []

with open("project_gutenberg.html", "r") as page:

    soup = BeautifulSoup(page, "html.parser")
    # print(soup.prettify())

    list_item_1 = soup.find("li")
    # print(list_item_1.prettify())

    book_items = soup.find_all("li", {"class":"booklink"})

    for list_item in book_items:

        title = list_item.find("span", {"class":"title"})
        author = list_item.find("span", {"class":"subtitle"})

        d = {"title":title, "author":author}
        books.append(d)

        print(title.text)
        print(author.text)
        print("****")
```

```python
with open ("../output/books_list.csv", mode="w") as csvfile:

    ## fieldnames have to be keys in our dictionary
    fieldnames = ["title", "author", "date"]

    writer = csv.DictWriter(csvfile, fieldnames=fieldnames)

    writer.writeheader()

    writer.writerows(books)
```
