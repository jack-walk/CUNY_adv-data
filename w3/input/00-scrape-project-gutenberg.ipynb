```python
import csv
from bs4 import BeautifulSoup
```

```python
with open("project_gutenberg.html", "r") as page:

    soup = BeautifulSoup(page, "html.parser")
    # print(soup.prettify())

    list_item_1 = soup.find("li")

    # print(list_item_1.prettify())

    book_items = soup.find_all("li")

    for book in book_items:
        print(book.text)
```
